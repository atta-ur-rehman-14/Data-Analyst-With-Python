```python
import pandas as pd
```

```python
data = {
    "name":       ["Alice", "Bob", "Carol", "David", "Eve"],
    "department": ["Engineering", "Finance", "Engineering", "HR", "Finance"],
    "salary":     [85000, 72000, 91000, 65000, 78000],
    "active":     [True, True, False, True, True]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df['department'].value_counts())
```

# Selecting Data

```python
print(df['department'])  # For a single column
```

```python
print(df[['department','salary']])  # # returns a DataFrame (one column, still a table)
```

# Selecting Rows — .loc[] and .iloc[]

### Pandas Indexing: `.loc[]` vs `.iloc[]`

| Method | Selection Basis | When to Use |
| :--- | :--- | :--- |
| **`.loc[]`** | Label (index value or column name) | You have meaningful index labels |
| **`.iloc[]`** | Integer position (`0, 1, 2...`) | You want rows by position |

```python
print(df.loc[2]) # Work on the basis of index numbers
```

```python
print(df.loc[0:3,"department"])  # by label
```

```python
print(df.loc[0:3,["department","active"]])
```

```python
# .iloc — by position
print(df.iloc[0])              # first row
```

```python
print(df.iloc[0:2, 0:2])       # rows 0-1, columns 0-1
```

# Filtering Rows with Conditions

```python
# Active Employees

Active = df[df["active"] == True]
print(Active)
```

```python
engineer = df[(df["department"] == "Engineering")]
print(engineer) 
                
```

```python
# Engineering department with salary above 80,000
senior_eng = df[(df["department"] == "Engineering") & (df["salary"] >= 80000)]
print(senior_eng)


```
