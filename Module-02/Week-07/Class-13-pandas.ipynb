# Series
- One Dimensional Array

```python
revenue = pd.Series([123,456,7889,986])
print(revenue)
```

```python
revenue = pd.Series([123,456,7889,986],
            index = ["jan","feb","march","april"])  # one dimensional array
print(revenue)
```

```python
print(revenue["jan"])
```

# DataFrames
- 2d with label raws ans column

```python
data = {
    "name":       ["Alice", "Bob", "Carol", "David", "Eve"],
    "department": ["Engineering", "Finance", "Engineering", "HR", "Finance"],
    "salary":     [85000, 72000, 91000, 65000, 78000],
    "active":     [True, True, False, True, True]
}

df = pd.DataFrame(data)
print(df)
```
