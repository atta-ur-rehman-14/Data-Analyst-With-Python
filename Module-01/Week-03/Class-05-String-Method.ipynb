# String Method
 - Raw string data almost never arrives clean. It comes in with extra spaces, inconsistent casing, and fields jammed together. Python's built-in string methods fix all of this.

```python
# Name 

# Ayan Hussain  -- 
# AYAN HUSSAIN 
# ayan hussain 
# AYAN hussain 
# ayan   hussain   -

# ayan hussain -- all names should be in this format 
```

### .Strip()

- Removes leading/trailing whitespace

```python
Name = "  Atta Ur Rehman "
print(Name)
Name = "Atta Ur Rehman"
print(Name)
```

```python
Name = "  Atta Ur Rehman "
print(Name)
print(Name.strip())
```

### .lstrip()

- Removes left whitespace only

### .rstrip()

-  Remove right spaces

```python
Name = "  Atta Ur Rehman "
print(Name,"---")
print(Name.strip())           # .strip()
print(Name.rstrip(),"---")    # .rstrip()
print(Name.lstrip(),"---")    # .lstrip()
```

### .Lower()
- All lowercase
### .Upper()
- All uppercase

```python
Institute = "Saylani welfare"

print(Institute.lower()) # .lower()
print(Institute.upper()) # .Upper()
```

### .title()
- First letter of each word capitalised
### .capitalize()
- First letter capitalized and other is small

```python
School = "center of exellence"
print(School.title())  # .title
print(School.capitalize()) # .capitalize
```

### .replace(a, b)
- Replaces all occurrences of a with b
### .split(x)
- Splits into a list on delimiter x
### .startswith(x)
- True if starts with x
### .endswith(x)
- True if ends with x

```python
# .replace()
Animal = "Cat"
Animal = Animal.replace("C","B")
print(Animal)
```

```python
# >split(x)
x = "Hello World"

print(x.split())  # sperated on the base of whitepace
                  
```

```python
x = "Hello, World, Jaranwala"

print(x.split('x'))  # separated on the base of ','.
```

```python
x = "Hello, World"

print(x.split('W'))  # separated on the base of 'W'
```
