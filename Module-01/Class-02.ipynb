## Logical Operators

- `And` --> Both true value is true otherwise false
- `OR` --> Both False value is False otherwise True
- `NOT` --> that reverses the Boolean value of a condition.
    - True → False
    - False → True

### And Operator

```python
Age = 22
Score = 90

print(Age > 18 and Score > 80)  # True
print(Age < 18 and Score > 80)  # False
print(Age > 18 and Score < 80)  # False
print(Age < 18 and Score < 80)  # False
```

### OR Operator

```python
Age = 22
Score = 90

print(Age < 18 or Score < 80)  # False
print(Age > 18 or Score < 80)  # True
print(Age < 18 or Score > 80)  # True
print(Age > 18 or Score > 80)  # True
```

### NOT Operator

```python
x = True
print(not x)  # False
```

```python
x = False
print(not x)  # True
```
