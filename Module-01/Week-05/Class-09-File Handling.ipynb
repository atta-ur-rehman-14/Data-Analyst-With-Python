```python
file = open("report.txt", "r") # File does not exist will you give an error
file.read()
file.close()
```

# `With` Statement

```python
with open(file = "report.txt" , mode = "r") as file:  # with -> automatically closes file when running the code is complete
    print(file.read())
```

```python
print(file.read())   # already closed with WITH statment
```

```python
# read
with open(file = "report.txt", mode = "r") as f:
    content = f.read()
    print(content)
```

# Readline
- readline - > return one line

```python
with open(file = "report.txt", mode = "r") as f:
    content = f.readline()
    print(content)
```

```python
with open(file = "report.txt", mode = "r") as f:
    content = f.readline(16) # read first 15 characters
    print(content)
```

# Readlines 
- read multiple lines

```python
with open(file = "report.txt", mode = "r") as f:
    content = f.readlines()
    print(content)

    print(content[0])
for line in content[0:3]:
    print(line)
```
