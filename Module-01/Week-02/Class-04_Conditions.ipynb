# Conditions

### If else

#### Syntax

```
if condition:
    Logic
else:
    Logic
```

````python
# If Age is greater or equal then 18 . Then you are Eligible For CNIC

Age = int(input("Enter the Age:"))
print(Age)

if Age >= 18:
    print("You are Eligible For CNIC.")
else:
    print("You are not Eligible For CNIC.")
````

````python
# Grading
# Marks >= 90 : Grade A+
# Marks >= 80 and < 90 : Grade A
# Marks >= 70 and < 80 : Grade B
# Marks >= 60 and < 70 : Grade C
# Marks >= 50 and < 60 : Grade D
# Marks < 50 : Grade F

Marks = int(input("Enter the Marks:"))
if Marks >= 90:
    print("Congratulations! You have scored an A+ Grade.")
elif Marks >= 80 and Marks < 90:
    print("You have scored an A Grade.")
elif Marks >= 70 and Marks < 80:
    print("You have scored a B Grade.")
elif Marks >= 60 and Marks < 70:
    print("You have scored a C Grade.")
elif Marks >= 50 and Marks < 60:
    print("You have scored a D Grade.")
else:
    print("You have scored an F Grade.")
````
