# Negative Indexing

```python
cities = ["Karachi", "Lahore", "Islamabad", "Peshawar"]
#            -4          -3          -2          -1     <- negative indices, counted from the end

# Negative indexing starts at -1 for the last element and counts backwards
# print(cities[3])

print(cities[-1])  # last element -> "Peshawar"
print(cities[-3])  # third from the end -> "Lahore"
```

### Append ()

- Add the new item at the end of the list

```python
cities = ["Faisalabd", "Lahore", "Karachi", "Islamadad", "Multan"]
numbers = [1,3,4,6,7]

cities.append("Peshawar")
numbers.append(8)
```

```python
print(cities)
print(numbers)
```

### POP ()

- removes and returns the last item. Pass an index to remove a specific item:

```python
cities.pop(2)   # remove with specific index
print(cities)
```

```python
cities.pop()  # removes the last element
print(cities)
```

```python
len(cities)
len(numbers)
```

```python
print(cities.index("Lahore"))
```

### Insert

```python
saylani_campuses = ['Z.A', 'Headoffice','Ibne-Adam']
# insert(index, value) -- inserts value at the given index, shifting later elements right

saylani_campuses.insert(0,"Malir-campus")
```

```python
print(saylani_campuses)
```

```python
saylani_campuses.insert(2, "Gulshan-campus")
print(saylani_campuses)
```

### Extend

- `extend()` is used with a list to add multiple elements from another iterable (usually another list) to the end of the list.

```python
saylani_campuses1 = ['Jehlum-campus', 'Z.A', 'Ibne-Adam', 'Malir-campus']
saylani_campuses2 = ['Lahore-campus', 'Faisalabad-campus', 'Gujranwala-campus']

saylani_campuses1.extend(saylani_campuses2)
```

```python
print(saylani_campuses1)
```

```python
saylani_campuses1[1] = 'Zaitoon Ashraf'
print(saylani_campuses1)
```

# Tuples

- A tuple looks like a list but uses parentheses instead of square brackets — and it cannot be changed after it is created.

```python
saylani_campuses = ('Lahore-campus', 'Faisalabad-campus', 'Gujranwala-campus')
print(type(saylani_campuses))
```

```python
saylani_campuses = ('Malir-campus', 'Z.A', 'lahore-campus', 'Headoffice', 'Ibne-Adam')
saylani_campuses[1] = 'Zaitoon Ashraf IT Park'  # tuples are immutable -- this raises TypeError
print(saylani_campuses)
print(type(saylani_campuses))
```
