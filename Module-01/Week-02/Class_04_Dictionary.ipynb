# Sets

- A set is a collection that stores unique values. Sets are especially useful for removing duplicates and quickly checking whether something exists in a collection.

````python
Fruits = {"Mango","Grapes","Banana"}

print(Fruits)
````

````python
numbers = {1, 2, 2, 3, 3, 3}

print(numbers)
````

````python
fruits = {"apple", "banana"}

fruits.add("mango")
fruits.remove("apple")

print(fruits)
````

## discard() 

- If the value you want to remove does not exist in the set, discard()  does not give an error.

````python
numbers = {10, 20, 30}

numbers.discard(50)

print(numbers)
````

# Dictionary

- A dictionary stores data as key-value pairs. Instead of accessing a value by position like a list, you access it by a name — its key.

````python
User = {
    "Name": "Sara",
    "Plan": "Pro",
    "Active" : True,
    "storage_gb": 250,
    "Marks" : [85, 90, 78]
}
````

````python
print(User["Plan"])
print(User.get("Name"))
print(User.get("Plan"))
````

````python
print(User["Marks"])
````

````python
#             key   indesx of list
print(User["Marks"][1])
````

````python
# "age" isn't a key in user -- direct [] access raises KeyError for a missing key
print(User["age"])
````

## get() Function

- To check key exist in the dicionary or not

````python
print(User.get("Age","N/A")) # return None if key is not present in the dictionary
print(User.get("Marks","N/A"))
````

### Update Dictionary

````python
User = {
    "Name": "Sara",
    "Plan": "Pro",
    "Active" : True,
    "storage_gb": 250,
    "Marks" : [85, 90, 78]
}

User["Name"] = "Jenny"   # # dicts are mutable -- assigning to an existing key updates its value
````

````python
print(User)
````

````python
User["Active"] = False  # same pattern: dict[key] = new_value
print(User)
````

### POP ()

````python
Remove = User.pop("Marks")
print(Remove)
````

````python
print(User)
````

````python
del User["storage_gb"]
print(User)
````

## Difference Between `pop()` and `del`

Both `pop()` and `del` are used to remove elements from a Python list.

| `pop()` | `del` |
|---|---|
| Removes an element from the list | Removes an element from the list |
| **Returns the removed element** | **Does not return the removed element** |
| We can store the removed value in a variable | We cannot get the deleted value from `del` |
| Can remove an element using its index | Can remove an element using its index |
| If no index is given, it removes the last element | An index or slice must be specified |
| Example: `x = numbers.pop(2)` | Example: `del numbers[2]` |

### Example

```python
numbers = [10, 20, 30, 40]

x = numbers.pop(2)
print(x)        # 30
print(numbers)  # [10, 20, 40]

del numbers[0]
print(numbers)  # [20, 40]
