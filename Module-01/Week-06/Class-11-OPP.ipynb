```python
student1 = {"name": "Ali",   "age": 22, "score": 85} # data
student2 = {"name": "Sara",  "age": 20, "score": 92}
student3 = {"name": "Ahmed", "age": 23, "score": 78}

def get_grade(score): # behaviour 
    for student in [student1, student2, student3]:
        if student["score"] == score:
            if score >= 90: return "A"
            elif score >= 80: return "B"
            else: return "C"

print(get_grade(85))
print(get_grade(92))
print(get_grade(78))
```

### Class
- A class is a blueprint. It describes what a thing holds (its attributes) and what it can do (its methods). An instance is one specific thing built from that blueprint.

- data --> attributes 
- functionality --> methods

```python
class Student:

    def __init__(self, name, age,score):
        self.name = name
        self.age = age
        self.score = score


# __init__ --> Constructor  no need to call -- will run automatically
# def __init__ — "when a new Student is created, do this:"
# self — "the new object itself"
# self.name = name — "store name on this object"

# object creation 
s3 = Student("Atta",50,836)
print(s3.name)
print(s3.age)
print(s3.score)

# Student.__init__(s3, "Ayan",50,836 )
#   s3.name = name
#   s3.age = age
#   s3.score =score
```

```python
class Student:

    def __init__(self, name, age,score): # data - attibutes
        self.name = name
        self.age = age
        self.score = score

    # methods 

    def get_grade(self):
        if self.score >= 90:
            return "A Grade"
        elif self.score >= 80:
            return "B Grade"
        else:
            return "C Grade"

Student_4 = Student("Mahad", 10, 95)
print(Student_4.name)

print(Student_4.get_grade())
```

```python
class student:
    def __init__ (self,Name,Roll_no,department,marks,total_marks):
        self.Name = Name
        self.Roll_no = Roll_no
        self.department = department
        self.marks = marks
        self.total_marks = total_marks
    def Calculate_percentage(self):
        percentage = self.marks / self.total_marks* 100
        return percentage
    def show_profile(self):
        print("Student Profile")
        print("Name:",self.Name)
        print("Roll-no:",self.Roll_no)
        print("department:",self.department)
        print("Marks:",self.marks)
        print("Percentage:",self.Calculate_percentage())

student1= student("Atta",27,"BSCS",450,500)
student1.show_profile()
```
