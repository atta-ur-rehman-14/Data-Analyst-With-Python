# Inheritance

- Inheritance lets a child class take on all the attributes and methods of a parent class, then add or change what it needs. Python supports five types.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age  = age

    def greet(self):
        print(f"Hi, I am {self.name}")

    def show_age(self):
        print(f"I am {self.age} years old")


class Teacher:
    def __init__(self, name, age):
        self.name = name           # ← same as Student
        self.age  = age            # ← same as Student

    def greet(self):               # ← same as Student
        print(f"Hi, I am {self.name}")

    def show_age(self):            # ← same as Student
        print(f"I am {self.age} years old")
```

```python
# Person  -- Parent Class
#   -- Student -- Child class
#   -- Teacher -- Child class


# Parent Class -- that being inherited
# Child class --  that inherits from parent class (Derived class)
```

```python
class Person:                # Parent Class
    def __init__(self, name, age):
        self.name = name
        self.age  = age

    def greet(self):
        return(f"Hi, I am {self.name}")

    def show_age(self):
        return(f"I am {self.age} years old")
class Student(Person):
    def study(self):
        return(f"{self.name} is studying!")
class Teacher(Person):
    def teach(self):
        return(f"{self.name} is teaching!")
Student1 = Student("Atta",22)
print(Student1.name)
print(Student1.greet())
print(Student1.show_age())
print(Student1.study())
print("***** Teacher *****")

Teacher1= Teacher("Ayan",25)
print(Teacher1.name)
print(Teacher1.greet())
print(Teacher1.show_age())
print(Teacher1.teach())
```

```python
class Animal:
    def __init__(self, name, sound):
        self.name  = name
        self.sound = sound

    def speak(self):
        return(f"{self.name} says {self.sound}")

    def sleep(self):
        return(f"{self.name} is sleeping...")


# Dog inherits everything from Animal
class Dog(Animal):
    def fetch(self):
        return(f"{self.name} fetches the ball!")

    def wag_tail(self):
        return(f"{self.name} is wagging its tail!")

Animal1= Animal("Cat","meoww")
print(Animal1.name)
print(Animal1.speak())
print(Animal1.sleep())
print("-----------------------")
Dog1 = Dog("Shera","Bark")
print(Dog1.name)
print(Dog1.fetch())
print(Dog1.wag_tail())
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age  = age

    def greet(self):
        return (f"Hi, I am {self.name}, {self.age} years old.")


class Student(Person):
    def __init__(self, name, age, student_id):
        # NOT using super() — setting manually
        self.name       = name        # repeat from Person
        self.age        = age         # repeat from Person
        self.student_id = student_id  # new attribute

    def show_id(self):
        return (f"Student ID: {self.student_id}")

    def greet(self): # override - replace parent's version
        print(f"My Name id {self.name},& Age is {self.age} years old.My Id is {self.student_id}")


student_1 = Student("Ayan",12,"S01")

print(student_1.show_id())
print(student_1.greet())
```

### With Super Function

```python
# With Super Function

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age  = age

    def greet(self):
        return (f"Hi, I am {self.name}, {self.age} years old.")


class Student(Person):
    def __init__(self, name, age, student_id):
        # using super() 
        super().__init__(name,age)
        self.student_id = student_id  # new attribute
    def show_id(self):
        return (f"Student ID: {self.student_id}")

    def greet(self): # override - replace parent's version
        return (f"My Name id {self.name},& Age is {self.age} years old.My Id is {self.student_id}")


student_1 = Student("Ayan",12,"S01")

print(student_1.show_id())
print(student_1.greet())
```

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def info(self):
        return(f"{self.brand} — {self.speed} km/h")


# WITHOUT super() — duplicate code
class Car(Vehicle):
    def __init__(self, brand, speed, doors):
        self.brand = brand    # copy from Vehicle
        self.speed = speed    # copy from Vehicle
        self.doors = doors


# WITH super() — clean, no duplication
class Car(Vehicle):
    def __init__(self, brand, speed, doors):
        super().__init__(brand, speed)   # Vehicle handles brand & speed
        self.doors = doors               # Car only adds what's new

# WITH super() — clean, no duplication
class Truck(Vehicle):
    def __init__(self, brand, speed, tires):
        super().__init__(brand, speed)   # Vehicle handles brand & speed
        self.tires = tires               # Car only adds what's new

Car1 = Car("Toyota",120,4)
print(Car1.info())
```
