# Task
- calculate grade of student s

- 90+ A+
-  80 - 90 -- A
- 70 - 80 B 
- 60 - 70 - C

```python
# define a function get_grade(score)
students = [
    {"name": "Ali", "score": 92},
    {"name": "Sara", "score": 75},
    {"name": "Ahmed", "score": 55},
    {"name": "Zara", "score": 88}
]

def get_grade(score):
    if score >= 90:
        return "A+"
    elif score < 90 and score >= 80:
        return "A"
    elif score < 80 and score >= 70:
        return "B"
    elif score < 70 and score >= 60:
        return "C"
    else:
        return "D"

for student in students:
    grade = get_grade(student["score"])
    print(f"{student['name']} got grade:{grade}")
```

# Task

- Build a receipt system using multiple functions that call each other.

- 1 function calculate_tax 
- 2 function calculate discount -- 8% 
- 3 function caluculate total 
- 4th function print_receipt - tax, dicount payment

- 10000 + 1000 -2000 = 9000

```python
def tax (amount, tax_rate):
    return amount * tax_rate
def calculate_discount(amount):
    return amount * 0.8
def Total_bill(amount, tax_rate):
    tax_amount = tax(amount, tax_rate)
    discount_amount = calculate_discount(amount)
    return amount + tax_amount - discount_amount
def Recipt(amount,tax_rate):
    tax_amount = tax(amount, tax_rate)
    total = Total_bill(amount, tax_rate)
    discount_amount = calculate_discount(amount)
    print(f"Amount: {amount}")
    print(f"Tax rate : {tax_amount}")
    print(f"Discount rate : {discount_amount}")
    print(f"Total Bill : {total}")
Recipt(10000, 0.10)
```

# Function Scope

```python

# scope means where a variable can be seen and used 

# local scope: variables created inside function - they only exists inside function, we can't call them outside 
```

```python
def sum(a,b):
    answer_1 =  a + b # variable initiated inside a function -- local scope
    return answer_1

sum(10,6)
print(answer_1)
```

```python
count = 10 

def increment():
    count = 100 # ik new variable banaya inside a function
    return count

increment()
print(count)
```

```python
count = 10 
# global scope

def increment():
    global count
    count = count + 100 # new variable initialize nhi kare ga 
    return count

increment()
print(count)
```
