1) Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

```python
import pandas as pd

# Create a DataFrame for 6 employees
df = pd.DataFrame({
    "Name": ["Ayesha", "Bilal", "Chen", "Deepa", "Ethan", "Farah"],
    "Age": [25, 32, 29, 41, 35, 28],
    "City": ["Delhi", "Lahore", "Delhi", "Karachi", "Lahore", "Delhi"],
    "Salary": [35000, 52000, 45000, 68000, 60000, 42000]
})

print("First 3 rows:")
display(df.head(3))
print("Shape:", df.shape)
print("Column names:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

2) Create a Series of 5 numbers with custom index labels ['a','b','c','d','e']. Print the value at label 'c', the values at positions 1 to 3, and the sum and mean of the Series.

```python
# Create a Series with custom index labels
numbers = pd.Series([10, 20, 30, 40, 50], index=["a", "b", "c", "d", "e"])

print("Value at label 'c':", numbers.loc["c"])
print("Values at positions 1 to 3:")
print(numbers.iloc[1:4])  # iloc uses an exclusive stop index
print("Sum:", numbers.sum())
print("Mean:", numbers.mean())
```

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
# Add Bonus and Total, rename City, then drop Bonus
df["Bonus"] = df["Salary"] * 0.10
df["Total"] = df["Salary"] + df["Bonus"]
df.rename(columns={"City": "Location"}, inplace=True)
df.drop(columns=["Bonus"], inplace=True)

print("Updated DataFrame:")
display(df)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
# Create a DataFrame with missing values
missing_df = pd.DataFrame({
    "Name": ["Ayesha", "Bilal", "Chen", "Deepa", "Ethan"],
    "Age": [25, None, 29, 41, None],
    "Salary": [35000, 52000, None, 68000, 60000],
    "City": ["Delhi", "Lahore", "Delhi", None, "Karachi"]
})

print("Missing values in each column:")
print(missing_df.isna().sum())

# Fill numeric missing values with each numeric column's mean
numeric_cols = missing_df.select_dtypes(include="number").columns
missing_df[numeric_cols] = missing_df[numeric_cols].fillna(missing_df[numeric_cols].mean())

# Drop any rows that still contain missing values (for example, missing City)
clean_df = missing_df.dropna()

print("\nDataFrame after filling numeric NaNs and dropping remaining missing rows:")
display(clean_df)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
# Sort by Salary descending, then Age ascending
sorted_df = df.sort_values(by=["Salary", "Age"], ascending=[False, True])
print("Sorted by Salary (descending), then Age (ascending):")
display(sorted_df)

print("Top 3 highest-paid employees:")
display(df.nlargest(3, "Salary"))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
# Group by City and calculate multiple statistics for Salary
city_summary = df.groupby("Location").agg(
    Average_Salary=("Salary", "mean"),
    Employee_Count=("Name", "count"),
    Maximum_Salary=("Salary", "max")
)
print("Salary summary by City:")
display(city_summary)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
# Create employees and departments DataFrames with a common DeptID
employees = pd.DataFrame({
    "EmployeeID": [1, 2, 3, 4],
    "Name": ["Ayesha", "Bilal", "Chen", "Deepa"],
    "DeptID": [10, 20, 10, 30]
})
departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "Department": ["Analytics", "Marketing", "Finance"]
})

print("Inner join (only matching DeptID values):")
display(pd.merge(employees, departments, on="DeptID", how="inner"))

print("Left join (all employees retained):")
display(pd.merge(employees, departments, on="DeptID", how="left"))

# Concatenate two DataFrames with the same structure, then remove duplicates
employees_part1 = employees.iloc[:3]
employees_part2 = pd.concat([employees.iloc[2:], employees.iloc[[0]]], ignore_index=True)
combined = pd.concat([employees_part1, employees_part2], ignore_index=True)
print("Concatenated DataFrame after removing duplicate rows:")
display(combined.drop_duplicates().reset_index(drop=True))
```
