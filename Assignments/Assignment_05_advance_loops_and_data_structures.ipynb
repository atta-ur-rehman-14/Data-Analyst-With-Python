## Question 1: Working with Lists

Create a list of integers, print every number, calculate the sum, and print only the even numbers.

```python
numbers = [12, 7, 24, 5, 18, 31, 42, 9, 16]

print("All numbers:")
total = 0
even_numbers = []


for number in numbers:
    print(number, end=" ")
    total += number

  
    if number % 2 == 0:
        even_numbers.append(number)

print(f"\nSum of all numbers: {total}")
print(f"Even numbers: {even_numbers}")
```

## Question 2: Dictionary Basics

The `student` dictionary contains a list of three marks. A loop is used both to display the dictionary and to calculate the average.

```python
student = {
    "name": "Ayesha",
    "age": 20,
    "course": "Computer Science",
    "marks": [78, 65, 82]
}

print("Student details:")
for key, value in student.items():
    print(f"{key}: {value}")


marks_total = 0
for mark in student["marks"]:
    marks_total += mark

average = marks_total / len(student["marks"])
result = "Passed" if average >= 50 else "Failed"
print(f"Average mark: {average:.2f}")
print(f"Result: {result}")
```

## Question 3: List of Dictionaries

Each employee is represented by a dictionary. A loop prints every employee, finds the highest salary, and calculates the total salary expense.

```python
employees = [
    {"name": "Ali", "department": "IT", "salary": 65000},
    {"name": "Sara", "department": "Marketing", "salary": 58000},
    {"name": "Hamza", "department": "Finance", "salary": 72000}
]

highest_paid = employees[0]
total_salary = 0

print("Employee details:")
for employee in employees:
    print(
        f"Name: {employee['name']}, "
        f"Department: {employee['department']}, "
        f"Salary: ${employee['salary']:,}"
    )

    total_salary += employee["salary"]
  
    if employee["salary"] > highest_paid["salary"]:
        highest_paid = employee

print(f"Highest-paid employee: {highest_paid['name']} (${highest_paid['salary']:,})")
print(f"Total salary expense: ${total_salary:,}")
```

## Question 4: While Loop with User Input

The function below asks for numbers until the user enters `-1`. It validates non-integer input and then displays the complete list, largest value, and smallest value.

A sample list is used for the automatic demonstration so the notebook runs without waiting for input. Call `collect_numbers()` with no argument when you want to enter numbers interactively.

```python
def collect_numbers(input_values=None):
    """Collect numbers until -1 is entered and report their range."""
    numbers_entered = []

    if input_values is None:

        while True:
            raw_value = input("Enter a number (-1 to stop): ")
            try:
                number = int(raw_value)
            except ValueError:
                print("Please enter a whole number.")
                continue

            if number == -1:
                break
            numbers_entered.append(number)
    else:
   
        for number in input_values:
            if number == -1:
                break
            numbers_entered.append(number)

    print(f"Complete list: {numbers_entered}")
    if numbers_entered:
        print(f"Largest number: {max(numbers_entered)}")
        print(f"Smallest number: {min(numbers_entered)}")
    else:
        print("No numbers were entered.")

    return numbers_entered


entered_numbers = collect_numbers([14, 3, 27, 8, -1])
```

## Question 5: Word Frequency Counter

The function splits a sentence into words and uses a dictionary to count each word. It converts words to lowercase so `Apple` and `apple` are counted together.

```python
def count_words(sentence=None):
    if sentence is None:
        sentence = input("Enter a sentence: ")

  
    words = sentence.lower().split()
    frequency = {}

   
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1

    print("Word frequency:")
    for word, count in frequency.items():
        print(f"{word}: {count}")

    return frequency


word_counts = count_words("apple banana apple orange banana apple")
```

## Question 6: Nested Loops — Multiplication Table

The outer loop selects each row and the inner loop calculates the five products in that row.

```python
print("Multiplication table from 1 to 5:")
for row in range(1, 6):
    row_values = []
    for column in range(1, 6):
        row_values.append(row * column)
    print(" ".join(f"{value:2}" for value in row_values))
```
